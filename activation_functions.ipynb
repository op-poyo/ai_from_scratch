```python
# activation functions are what allow nn to not be purely linear
# if you think about the prev notebook, it was effectively a linear regression algorithim
# finding a weight and a bias, which are m and c in y = mx + c

# even with multiple inputs, x1m1 + x2m2 + c =  y is still fundamentally linear
# so we add a bit of non linearaity between each layer
```

```python
import math
import numpy
```

```python
def binary_step(x):
    return 0 if x < 0 else 1
print(binary_step(10))
print(binary_step(0))
print(binary_step(-1))

# this one kinda sucks but serves as a simple idea: this is some non linear behaviour added into the mix
```

```python
def sigmoid(x):
    value = 1 / (1+math.e**-x)
    return value
def tanh(x):
    value = (math.e**x - math.e**-x)/(math.e**x + math.e**-x)
    return value
def ReLU(x):
    return (max(x,0))
def lRelu(x):
    return(max(x*0.1,x))
def pRelu(x,a):
    return(max(a*x,x))
```

```python
# nn structure so far
# input -> layer 1 ->  activation function ..... -> output layer
# now the class model from before was easy to understand but still quite slow

# instead we can rewrite it in terms of matrixes
```

```python
# for backpropogation, derivatives will be required

def sigmoid_derivative(output):
    value = output * (1-output)
    return value
def tanh_derivative(output):
    value = 1.0 - (output ** 2)
    return value
def Relu_derivative(output):
    value = 1 if output > 0 else 0
    return value
def lRelu_derivative(output):
    value = 1 if output>0 else 0.1
    return value
def pRelu_derivative(output, a):
    value = 1 if output > 0 else a
    return value
```
