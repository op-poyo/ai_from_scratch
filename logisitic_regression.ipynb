```python
import math
import random

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def binary_cross_entropy(actual,predicted):
    value = -(actual * math.log(predicted) +
        (1 - actual) * math.log(1 - predicted))
    return value

class Neuron:
    def __init__(self, weights, bias):
        self.weights = weights
        self.bias = bias

def predicted(x,neuron):
    first = x*neuron.weight + neuron.bias
    second = sigmoid(first)
    return second


def generate_pass_fail(n=1000):
    data = []
    for _ in range(n):
        hours = random.uniform(0, 10)
        sleep = random.uniform(4, 9)
        z = 0.8*hours + 0.5*sleep - 7
        p = sigmoid(z)
        y = random.random() < p
        data.append(((hours, sleep), int(y)))
    return data

def forward_pass(x, neuron):
    z = 0
    for xi, wi in zip(x, neuron.weights):
        z += xi * wi
    z += neuron.bias

    y_hat = sigmoid(z)
    return y_hat
```

```python

data = generate_pass_fail(5)

neuron = Neuron(weights=[0.1, -0.2], bias=0.0)

for x, y in data:
    y_hat = forward_pass(x, neuron)
    loss = binary_cross_entropy(y, y_hat)
    print(f"x={x}, y={y}, ŷ={y_hat:.3f}, loss={loss:.3f}")
```

```python
# dL/dy^ = y^ - y / y^(1-y)
# dy^/dz = y^(1-y)
learning_rate = 0.1
error = y_hat - y  # dL/dz

# update weights
for i in range(len(neuron.weights)):
    neuron.weights[i] -= learning_rate * error * x[i]

# update bias
neuron.bias -= learning_rate * error
```

```python
def train(neuron, data, lr=0.01, epochs=100):
    for epoch in range(epochs):
        total_loss = 0
        for x, y in data:
            # forward
            y_hat = forward_pass(x, neuron)
            
            # compute loss
            loss = binary_cross_entropy(y, y_hat)
            total_loss += loss
            
            # backprop
            error = y_hat - y
            for i in range(len(neuron.weights)):
                neuron.weights[i] -= lr * error * x[i]
            neuron.bias -= lr * error
        
        avg_loss = total_loss / len(data)
        if epoch % 10 == 0:
            print(f"Epoch {epoch}: avg loss = {avg_loss:.4f}")
```

```python
data = generate_pass_fail(500)
neuron = Neuron(weights=[0.0, 0.0], bias=0.0)

train(neuron, data, lr=0.01, epochs=500)
```

```python
# make predictions after training
for x, y in data[:5]:
    y_hat = forward_pass(x, neuron)
    print(f"x={x}, y={y}, ŷ={y_hat:.3f}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

X = torch.tensor([[1., 2.],
                  [2., 1.],
                  [2., 3.],
                  [3., 2.]])
y = torch.tensor([[0.], [0.], [1.], [1.]])

model = nn.Linear(2, 1)
criterion = nn.BCEWithLogitsLoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)

for _ in range(1000):
    optimizer.zero_grad()

    logits = model(X)
    loss = criterion(logits, y)

    loss.backward()
    optimizer.step()

print(model.weight, model.bias)
```
